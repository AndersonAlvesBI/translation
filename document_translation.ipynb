```python
import requests
from docx import Document
import os
subscription_key = "9hJpcfPpOeWSNxrsuSLxHGdJUOhS8XBXzNQcNoOpMN6DMCezvdqaJQQJ99BJACHYHv6XJ3w3AAAbACOG22ez"
endpoint = 'https://api.cognitive.microsofttranslator.com'
location = "eastus2"
language_destination = 'pt-br'

def translator_text(text, target_language):
  path = '/translate'
  constructed_url = endpoint + path

  headers = {
      'Ocp-Apim-Subscription-Key': subscription_key,
      'Ocp-Apim-Subscription-Region': location,
      'Content-type': 'application/json',
      'X-ClientTraceId': os.urandom(16).hex()
  }

  body = [{
      'text': text
  }]

  params = {
      'api-version': '3.0',
      'from': 'en',
      'to': language_destination
  }

  request = requests.post(constructed_url, params=params, headers=headers, json=body)
  response = request.json()
  return response[0]["translations"][0]["text"]
```

```python
def translator_document(path):
  doc = Document(path) # Corrige a criação do objeto Document
  full_text = []

  print("\n🔹 Traduzindo o documento, aguarde...\n")

  # Traduz parágrafo por parágrafo
  for paragraph in doc.paragraphs:
    if paragraph.text.strip(): # ignora parágrafos vazios
      translated_text = translator_text(paragraph.text, language_destination)
      print(f"🟢 Original: {paragraph.text}")
      print(f"🔵 Traduzido: {translated_text}\n")
      full_text.append(translated_text)
    else:
      full_text.append("")

  # Cria novo documento traduzido
  translated_doc = Document()
  for line in full_text:
    translated_doc.add_paragraph(line)

  # Define caminho de saída
  path_translated = path.replace('.docx', f"_{language_destination}.docx")
  translated_doc.save(path_translated)

  print(f"\n✅ Tradução concluída! Arquivo salvo em: {path_translated}")
  return path_translated

# Exemplo de uso:
if __name__ == "__main__":
  input_file = "/content/musica.docx"
  translator_document(input_file)
```
